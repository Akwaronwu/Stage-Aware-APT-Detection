```python
# ============================================================
# PHASE 1 — DATASET LOADING & VERIFICATION
# NF-UQ-NIDS-v2
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# 1. Dataset path
# ------------------------------------------------------------

DATA_PATH = Path(r"C:\Users\HomePC\Documents\Research\Dataset\NF-UQ-NIDS-v2\Complete_set\data\NF-UQ-NIDS-v2.csv")

# Verify that the file exists
if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at:\n{DATA_PATH}\n\n"
        "Please update DATA_PATH to the correct CSV location."
    )

print("Dataset found:")
print(DATA_PATH)
print(f"File size: {DATA_PATH.stat().st_size / (1024**2):.2f} MB")
```

```python
# ------------------------------------------------------------
# 2. Inspect column names
# ------------------------------------------------------------

header_df = pd.read_csv(
    DATA_PATH,
    nrows=5
)

print("Number of columns:", len(header_df.columns))
print("\nColumn names:")
for i, col in enumerate(header_df.columns, start=1):
    print(f"{i:02d}. {col}")
```

```python
# ------------------------------------------------------------
# 3. Preview the dataset
# ------------------------------------------------------------

print("\nFirst five records:")
display(header_df)
```

```python
# ------------------------------------------------------------
# 4. Identify key columns
# ------------------------------------------------------------

print("\nImportant columns detected:")

for target_col in ["Label", "Attack", "Dataset"]:
    matches = [col for col in header_df.columns if col.strip().lower() == target_col.lower()]

    if matches:
        print(f"{target_col}: {matches}")
    else:
        print(f"{target_col}: NOT FOUND")
```

```python
# ------------------------------------------------------------
# 5. Count total rows safely
# ------------------------------------------------------------

CHUNK_SIZE = 100_000

total_rows = 0

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=CHUNK_SIZE,
    usecols=[header_df.columns[0]]
):
    total_rows += len(chunk)

print(f"\nTotal rows: {total_rows:,}")
print(f"Total columns: {len(header_df.columns):,}")
print(f"Dataset shape: ({total_rows:,}, {len(header_df.columns)})")
```

```python
# ------------------------------------------------------------
# 6. Inspect data types
# ------------------------------------------------------------

print("\nData types:")
print(header_df.dtypes)
```

```python
# ------------------------------------------------------------
# 7. Locate Label and Attack columns
# ------------------------------------------------------------

def find_column(columns, target):
    matches = [
        col for col in columns
        if col.strip().lower() == target.lower()
    ]

    if len(matches) == 0:
        return None

    if len(matches) > 1:
        raise ValueError(
            f"Multiple columns match '{target}': {matches}"
        )

    return matches[0]


LABEL_COL = find_column(header_df.columns, "Label")
ATTACK_COL = find_column(header_df.columns, "Attack")
DATASET_COL = find_column(header_df.columns, "Dataset")

print("\nDetected columns:")
print("Label   :", LABEL_COL)
print("Attack  :", ATTACK_COL)
print("Dataset :", DATASET_COL)
```

```python
# ------------------------------------------------------------
# 8. Label distribution
# ------------------------------------------------------------

if LABEL_COL is None:
    raise ValueError(
        "The 'Label' column was not found. "
        "Check the column names printed above."
    )

label_counts = {}

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=CHUNK_SIZE,
    usecols=[LABEL_COL]
):
    counts = chunk[LABEL_COL].value_counts(dropna=False)

    for value, count in counts.items():
        label_counts[value] = label_counts.get(value, 0) + count

label_distribution = (
    pd.Series(label_counts, name="Instances")
    .sort_values(ascending=False)
)

print("\nLabel Distribution:")
display(label_distribution.to_frame())
```

```python
# ------------------------------------------------------------
# 9. Label percentages
# ------------------------------------------------------------

label_summary = pd.DataFrame({
    "Instances": label_distribution,
    "Percentage": (
        label_distribution / label_distribution.sum() * 100
    )
})

label_summary["Percentage"] = label_summary["Percentage"].round(4)

print("\nLabel Summary:")
display(label_summary)
```

```python
# ------------------------------------------------------------
# 10. Attack-category distribution
# ------------------------------------------------------------

if ATTACK_COL is None:
    raise ValueError(
        "The 'Attack' column was not found. "
        "Check the column names printed above."
    )

attack_counts = {}

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=CHUNK_SIZE,
    usecols=[ATTACK_COL]
):
    counts = chunk[ATTACK_COL].value_counts(dropna=False)

    for value, count in counts.items():
        attack_counts[value] = attack_counts.get(value, 0) + count

attack_distribution = (
    pd.Series(attack_counts, name="Instances")
    .sort_values(ascending=False)
)

print("\nAttack Category Distribution:")
display(attack_distribution.to_frame())
```

```python
# ------------------------------------------------------------
# 11. Attack-category percentages
# ------------------------------------------------------------

attack_summary = pd.DataFrame({
    "Instances": attack_distribution,
    "Percentage": (
        attack_distribution / attack_distribution.sum() * 100
    )
})

attack_summary["Percentage"] = attack_summary["Percentage"].round(4)

print("\nAttack Category Summary:")
display(attack_summary)
```

```python
# ------------------------------------------------------------
# 12. Expected categories from the study's Table 1
# ------------------------------------------------------------

EXPECTED_ATTACK_CATEGORIES = [
    "Reconnaissance Scanning",
    "Reconnaissance Activities",
    "Password / Brute Force Attacks",
    "Exploit Attacks",
    "Malicious Code Execution",
    "Backdoor Bot Activities",
    "Backdoor Attacks",
    "Internal Compromise Activities",
    "Worm Propagation",
    "Man-in-the-Middle (MITM) Attacks",
    "Unauthorized Data Extraction",
    "Ransomware Attacks"
]

observed_categories = set(
    str(x).strip()
    for x in attack_distribution.index
    if pd.notna(x)
)

expected_categories = set(EXPECTED_ATTACK_CATEGORIES)

missing_from_dataset = expected_categories - observed_categories
unexpected_categories = observed_categories - expected_categories

print("\n" + "=" * 60)
print("CATEGORY VERIFICATION")
print("=" * 60)

print("\nExpected categories found:")
for category in EXPECTED_ATTACK_CATEGORIES:
    status = "FOUND" if category in observed_categories else "NOT FOUND"
    print(f"{status:12} : {category}")

print("\nCategories in dataset but NOT in Table 1:")
if unexpected_categories:
    for category in sorted(unexpected_categories):
        print(" -", category)
else:
    print("None")
```

```python
# ------------------------------------------------------------
# 13. Dataset metadata column
# ------------------------------------------------------------

if DATASET_COL is not None:

    print("\n'Dataset' column detected.")

    dataset_values = {}

    for chunk in pd.read_csv(
        DATA_PATH,
        chunksize=CHUNK_SIZE,
        usecols=[DATASET_COL]
    ):
        counts = chunk[DATASET_COL].value_counts(dropna=False)

        for value, count in counts.items():
            dataset_values[value] = (
                dataset_values.get(value, 0) + count
            )

    dataset_summary = pd.Series(
        dataset_values,
        name="Instances"
    ).sort_values(ascending=False)

    display(dataset_summary.to_frame())

else:
    print("\n'Dataset' column was not found.")
```

```python
# ------------------------------------------------------------
# 14. PHASE 1 VERIFICATION REPORT
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("PHASE 1 — NF-UQ-NIDS-v2 DATASET VERIFICATION REPORT")
print("=" * 70)

print(f"Dataset file       : {DATA_PATH.name}")
print(f"File size          : {DATA_PATH.stat().st_size / (1024**2):.2f} MB")
print(f"Total observations : {total_rows:,}")
print(f"Total features     : {len(header_df.columns):,}")

print("\nKey columns")
print("-" * 70)
print(f"Label column       : {LABEL_COL}")
print(f"Attack column      : {ATTACK_COL}")
print(f"Dataset column     : {DATASET_COL}")

print("\nLabel distribution")
print("-" * 70)
print(label_summary)

print("\nAttack categories")
print("-" * 70)
print(attack_summary)

print("\nExpected categories missing")
print("-" * 70)

if missing_from_dataset:
    for category in sorted(missing_from_dataset):
        print(" -", category)
else:
    print("None")

print("\nUnexpected categories")
print("-" * 70)

if unexpected_categories:
    for category in sorted(unexpected_categories):
        print(" -", category)
else:
    print("None")

print("\n" + "=" * 70)
print("PHASE 1 VERIFICATION COMPLETE")
print("=" * 70)
```

```python
# ------------------------------------------------------------
# Missing-value and infinite-value analysis
# ------------------------------------------------------------

CHUNK_SIZE = 100_000

missing_counts = None
infinity_counts = None
total_rows = 0

for chunk in pd.read_csv(
    DATA_PATH,
    chunksize=CHUNK_SIZE,
    low_memory=True
):

    total_rows += len(chunk)

    # --------------------------------------------------------
    # Missing values
    # --------------------------------------------------------

    chunk_missing = chunk.isna().sum()

    if missing_counts is None:
        missing_counts = chunk_missing
    else:
        missing_counts = missing_counts.add(
            chunk_missing,
            fill_value=0
        )

    # --------------------------------------------------------
    # Infinite values
    # --------------------------------------------------------

    numeric_columns = chunk.select_dtypes(
        include=[np.number]
    ).columns

    chunk_infinity = np.isinf(
        chunk[numeric_columns]
    ).sum()

    if infinity_counts is None:
        infinity_counts = chunk_infinity
    else:
        infinity_counts = infinity_counts.add(
            chunk_infinity,
            fill_value=0
        )


# ------------------------------------------------------------
# Missing-value percentages
# ------------------------------------------------------------

missing_percentage = (
    missing_counts / total_rows * 100
).round(4)

missing_table = pd.DataFrame({
    "Missing Values": missing_counts.astype(int),
    "Percentage": missing_percentage
})

missing_table = missing_table[
    missing_table["Missing Values"] > 0
].sort_values(
    "Missing Values",
    ascending=False
)

print("Missing-value summary:")

if missing_table.empty:
    print("✓ No missing values detected.")
else:
    display(missing_table)


# ------------------------------------------------------------
# Infinite-value analysis
# ------------------------------------------------------------

infinity_summary = (
    infinity_counts[infinity_counts > 0]
    .sort_values(ascending=False)
)

print("\nInfinite-value summary:")

if infinity_summary.empty:
    print("✓ No infinite values detected.")
else:
    display(
        infinity_summary.to_frame("Infinite Values")
    )
```
